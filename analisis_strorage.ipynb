```python

# 1. Data arsip dan asumsi ukuran
data_arsip = [
    {"jenis": "Dokumen teks", "format": "TXT, CSV, JSON",
     "jumlah": 2000, "ukuran": 2 / 1024, "satuan": "KB"},
    {"jenis": "Foto dokumentasi", "format": "JPG, PNG",
     "jumlah": 1000, "ukuran": 3, "satuan": "MB"},
    {"jenis": "Rekaman audio", "format": "MP3, WAV",
     "jumlah": 100, "ukuran": 20, "satuan": "MB"},
    {"jenis": "Video pertunjukan", "format": "MP4, MOV",
     "jumlah": 100, "ukuran": 500, "satuan": "MB"},
    {"jenis": "Model 3D", "format": "OBJ, PLY, STL",
     "jumlah": 20, "ukuran": 50, "satuan": "MB"},
    {"jenis": "Data lokasi", "format": "GeoJSON",
     "jumlah": 10, "ukuran": 1 / 1024, "satuan": "KB"}
]

# 2. Tampilkan karakteristik dan hitung ukuran total
print("=" * 75)
print("RENCANA AKUISISI DATA ARSIP DIGITAL SANGGAR TARI")
print("=" * 75)

total_mb = 0

for data in data_arsip:
    ukuran_total_mb = data["jumlah"] * data["ukuran"]
    total_mb += ukuran_total_mb

    print(f"\nJenis data       : {data['jenis']}")
    print(f"Format           : {data['format']}")
    print(f"Jumlah data      : {data['jumlah']} unit")
    print(f"Ukuran per unit  : {data['ukuran']:.4f} {data['satuan']}")
    print(f"Total ukuran     : {ukuran_total_mb / 1024:.4f} GB")

# 3. Hitung total penyimpanan dan cadangan 30%
cadangan_mb = total_mb * 0.30
total_dengan_cadangan_mb = total_mb + cadangan_mb

print("\n" + "=" * 75)
print("HASIL ESTIMASI PENYIMPANAN")
print("=" * 75)
print(f"Total kebutuhan awal : {total_mb / 1024:.2f} GB")
print(f"Cadangan 30%         : {cadangan_mb / 1024:.2f} GB")
print(f"Total yang disiapkan : {total_dengan_cadangan_mb / 1024:.2f} GB")
```
